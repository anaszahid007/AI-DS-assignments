#### Challenge 2: Building & Manipulating Pandas DataFrames
- create a total_value column
- multiply the price with the quantity of the product store it there parallel total_value column
- sum the totla prices to show at the last result 



```python
import pandas as pd
import numpy as np

data = {
    "product": ["Laptop", "Mouse", "Keyboard", "Monitor"],
    "price": [120000, 2500, 5000, 35000],
    "quantity": [2, 10, 5, 3],
}

df = pd.DataFrame(data)
df["total_value"] = df["price"] * df["quantity"]
total_price = np.sum(df["total_value"])

print(df)
print("\nOverall Inventory Value (PKR):", total_price)
```

#### Challenge 3: Filtering Pandas DataFrames with Single Conditions
- find who is salary greater than 90000 
- find those employes who is in AI dempartment

```python
import pandas as pd
import numpy as np

data = {
    "name": ["Ayesha", "Ali", "Bilal", "Sara", "Zaid"],
    "department": ["AI", "Web Dev", "AI", "Data Science", "Web Dev"],
    "salary": [120000, 85000, 150000, 95000, 70000],
}

df = pd.DataFrame(data)

high_earners = df[df["salary"] > 90000]
belong_to_ai_department = df[df["department"] == "AI"]

print(high_earners)
print("\n")
print(belong_to_ai_department)
```

#### Challenge 4: Combining Multiple Conditions
1. Extract and print employees who work in "Web Dev" OR earn a salary less than 80,000 PKR.

2. Extract and print employees who work in "AI" AND earn a salary greater than 130,000 PKR.


```python
import pandas as pd

data = {
    "name": ["Ayesha", "Ali", "Bilal", "Sara", "Zaid"],
    "department": ["AI", "Web Dev", "AI", "Data Science", "Web Dev"],
    "salary": [120000, 85000, 150000, 95000, 70000]
}

df = pd.DataFrame(data)

webdev_dep_less_sal = df[(df["department"] == "Web Dev") | (df["salary"] < 80000)]
ai_dep_gret_sal = df[(df["department"] == "AI") & (df["salary"] > 130000)]

print(webdev_dep_less_sal)
print("\n")
print(ai_dep_gret_sal)
```

#### Challenge 7: Grouping & Aggregations with groupby()

```python
import pandas as pd

data = {
    "name": ["Ayesha", "Ali", "Bilal", "Sara", "Zaid", "Usman"],
    "department": ["AI", "Web Dev", "AI", "Data Science", "Web Dev", "Data Science"],
    "salary": [120000, 85000, 150000, 95000, 70000, 110000]
}

df = pd.DataFrame(data)
average_salary = df.groupby("department")["salary"].mean()
total_salary = df.groupby("department")["salary"].sum()
print(average_salary)
print("\n")
print(total_salary)
```

#### Challenge 8: Handling Missing Data (NaN)
1. Check the count of missing (NaN) values in each column using df.isna().sum().
2. Fill all missing values in the "price" column with the number 0 using .fillna().
3. Drop any rows where the "sales" column has a missing (NaN) value using .dropna(subset=["sales"]).

```python
import pandas as pd
import numpy as np

data = {
    "product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"],
    "price": [120000, np.nan, 5000, 35000, np.nan],
    "sales": [15, 50, np.nan, 20, 30]
}

df = pd.DataFrame(data)


total_missing_values = df["price"].isna().sum()
df['price'] = df['price'].fillna(0)
df_sales_clean = df.dropna(subset=["sales"])
across_na = df.isna().sum()


print("DataFrame")
print(df)
print("\n")

print("Total missing values: ", total_missing_values)
print("\n")

print("Sales null Clean")
print(df_sales_clean)
print("\n")

print("Across NULL")
print(across_na)
```

#### Challenge 9: Sorting & Selecting Data
1. Sort df by "price" in descending order (highest price at the top) and store it in df_sorted.

2. Get and print the top 2 most expensive products using df_sorted.head(2).

```python
import pandas as pd

data = {
    "product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"],
    "price": [120000, 2500, 5000, 35000, 8000],
    "units_sold": [15, 50, 40, 20, 30]
}

df = pd.DataFrame(data)

df_store = df.sort_values(by="price", ascending=False)
top_price = df_store.head(2)

print(df_store)
print("\n")
print(top_price)
```

#### Challenge 10: Applying Custom Transformations (.apply() & lambda)
1. Create a new column named "category" using .apply():
    - Label items as "High End" if the price is greater than or equal to 30000.
    - Label items as "Budget" if the price is less than 30000.

2. Print the updated DataFrame df.

```python
import pandas as pd

data = {
    "product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"],
    "price": [120000, 2500, 5000, 35000, 8000]
}

df = pd.DataFrame(data)

df['category'] = df['price'].apply(lambda x: "Highe End" if x >= 30000 else "Budget")

print(df)
```

#### Challenge 11: Merging DataFrames with pd.merge()
1. Combine df_info and df_sales into a single DataFrame named df_merged by matching their "product_id" column.

2. Print df_merged.

```python
import pandas as pd

df_info = pd.DataFrame({
    "product_id": [101, 102, 103, 104],
    "product_name": ["Laptop", "Mouse", "Keyboard", "Monitor"]
})

df_sales = pd.DataFrame({
    "product_id": [101, 102, 103, 104],
    "units_sold": [12, 45, 30, 18]
})

df_combined = pd.merge(df_info, df_sales, on="product_id")

df_merge = pd.DataFrame(df_combined)
print(df_merge)
```
