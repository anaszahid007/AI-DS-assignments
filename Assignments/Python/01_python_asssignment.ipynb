6. Rectangle Area:

### 7. Time Converter:
Store 250 minutes in a variable. Calculate and print the number of hours and remaining minutes. 

```python
total_minutes = 250
hours = total_minutes // 60
remaining_minutes = total_minutes % 60
print(f"Hours: {hours}")
print(f"Minutes: {remaining_minutes}")
```

### 8. Student Eligibility:
Create variables for marks and attendance.
- A student is eligible if: 
- Marks are greater than or equal to 50  
- Attendance is greater than or equal to 75  Use logical operators to calculate and print whether the student is eligible. 

```python
marks = 82
attendance = 80
is_eligible = (marks >= 50) and (attendance >= 75)
print(f"Is the student eligible? {is_eligible}")
```
